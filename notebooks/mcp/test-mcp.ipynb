# MCP retrieval experiments

Run through `scripts/slurm_qwen_mcp.sh`. Full results and traces are saved under `rdfsolve/logs/mcp-test`. Tables below display ten rows.

```python
from pathlib import Path
import os
from rdfsolve.api import ask_rdf

root = Path(os.environ["RDFSOLVE_ROOT"])
schema = Path(os.environ.get("RDFSOLVE_SCHEMA", root / "notebooks/mcp/schemas/aopwikirdf.schema.json"))
```

```python
questions = [
    "List the Adverse Outcome Pathways in AOP-Wiki.",
    "List the Adverse Outcome Pathways in AOP-Wiki that are related to thyroid in humans, and the associated chemical stressors with all their available names and IDs.",
    "Return Key Events with their applicable species and measurement methods for all Adverse outcome pathways in human taxon",
]

answers = []
for question in questions:
    print("QUESTION:", question)
    answer = await ask_rdf(question, schema=schema, source_id="aopwikirdf")
    answers.append(answer)
    print(answer.text)
    print(answer.diagnostics())
    print(answer.query)
    if answer.state == "complete":
        display(answer.table().head(10))
```

```python
assert all(a.state == "complete" for a in answers), "See saved answer artifacts for failed investigations."
```
