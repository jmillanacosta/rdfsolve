# AOPWiki through the package-first MCP

Use the existing running model and SLURM setup. No model server or schema mining
starts in this notebook. `QWEN_*` and `RDFSOLVE_MODEL*` settings are inherited.
The schema remains `notebooks/data/aopwikirdf.schema.json` regardless of working
directory. Explicit `rdf_finish` ends the agent; a probe is not a final answer.
These are manual smoke questions, not an F1 experiment or certified gold answers.


```python
from rdfsolve.openai import ask_aopwiki
from IPython.display import display
import json
```

```python
questions = ['List the Adverse Outcome Pathways in AOP-Wiki.', 'List the Adverse Outcome Pathways in AOP-Wiki that are related to thyroid in humans, and the associated chemical stressors with all their available names and IDs.', 'Return Key Events with their applicable species and measurement methods for all Adverse outcome pathways in human taxon']
answers = []
```

```python
for question in questions:
    print("\nQUESTION:", question)
    answer = await ask_aopwiki(question)
    answers.append(answer)
    print(answer.text)
    print(json.dumps(answer.diagnostics(), indent=2, default=str))
    print(answer.query)
    if answer.state == "complete":
        display(answer.table())
    else:
        print(answer.error)
```

`answer.package` records actual client-method counts, timings and source queries.
`answer.calls` records the model's tool actions. `answer.bindings` contains the full
final data, not the preview. `answer.save("last_answer.json")` retains the result.
When `RDFSOLVE_QWEN_OUTPUT` is set by the existing job, diagnostics save automatically.
No reviewer or natural-language coverage certificate is produced.
