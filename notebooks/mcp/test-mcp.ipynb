# Ask AOPWiki with a local Qwen model

Using the Qwen checkpoint served by llama.cpp on the HPC.

Set `QWEN_BASE_URL` and `QWEN_MODEL` to compare another installed Qwen checkpoint.

```python
from rdfsolve.openai import ask_aopwiki
import json
import pandas as pd
from IPython.display import display
from pydantic_ai.messages import ModelResponse, RetryPromptPart


answer = await ask_aopwiki("List the Adverse Outcome Pathways in AOP-Wiki.")
print(answer.text)
display(answer.table())
```

```python
usage = answer.usage() if callable(answer.usage) else answer.usage
messages = answer.run.new_messages()
responses = [m for m in messages if isinstance(m, ModelResponse)]
repairs = [
    p for m in messages for p in m.parts
    if isinstance(p, RetryPromptPart)
]

print("State:", answer.state)
print("Retrieved rows:", len(answer.bindings) if answer.state == "complete" else "not executed")
print("Model requests:", usage.requests)
print("Reported tokens:", {
    "input": usage.input_tokens,
    "output": usage.output_tokens,
    "total": usage.total_tokens,
})
print("Extra usage details:", usage.details)
print("Recorded MCP calls:", len(answer.calls), "| Repair prompts:", len(repairs))
print("Schema:", answer.schema)

# Context growth and generation size for each model response.
display(pd.DataFrame([
    {
        "request": i,
        "model": m.model_name,
        "input_tokens": m.usage.input_tokens,
        "output_tokens": m.usage.output_tokens,
        "cache_read_tokens": m.usage.cache_read_tokens,
        "finish_reason": m.finish_reason,
    }
    for i, m in enumerate(responses, 1)
]))

# Resolve selected handles against options actually offered by the package.
options, rows = {}, []
for i, call in enumerate(answer.calls, 1):
    args, result = call["arguments"], call["result"]
    action = args.get("action") or {}
    selected = options.get(
        (args.get("session_id"), action.get("decision"), action.get("option")),
        action.get("option", ""),
    )
    decision = result.get("decision") or {}
    for option in decision.get("options", []):
        options[(result.get("session"), decision["id"], option["id"])] = option["meaning"]

    rows.append({
        "step": i,
        "tool": call["name"],
        "action_or_target": action.get("type", args.get("target", "")),
        "state": result.get("state", ""),
        "revision": result.get("revision"),
        "choice_requested": selected,
        "offered": " | ".join(o["meaning"] for o in decision.get("options", [])),
        "result_JSON_bytes": len(
            json.dumps(result, ensure_ascii=False, separators=(",", ":")).encode("utf-8")
        ),
        "issue": result.get("error") or result.get("message") or result.get("reason", ""),
    })

trace = pd.DataFrame(rows)
with pd.option_context("display.max_rows", None, "display.max_colwidth", 100):
    display(trace)

if repairs:
    display(pd.DataFrame([
        {"tool": p.tool_name or "output validation", "reason": p.content}
        for p in repairs
    ]))

print("Executed SPARQL:\n", answer.query or "No executed query retained.")
```
