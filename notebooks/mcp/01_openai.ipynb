# Ask about AOPWiki

Use the model configured for this notebook. Local models run through the SLURM launcher.

```python
from rdfsolve.api import ask_rdf

question = "Which pathways are linked to Phenobarbital?"
```

```python
answer = await ask_rdf(
    question,
    schema="../data/aopwikirdf.schema.json",
    max_response_tokens=4096,
)
print(answer.text)
```

## Read the answer

```python
answer.table()
```

## See the query and what happened

```python
print(answer.query)
answer.diagnostics()
```
