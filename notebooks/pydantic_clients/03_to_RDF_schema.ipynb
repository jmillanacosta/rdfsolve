# Turn the chemical table into RDF

Use the chemical names and identifiers saved in the [thyroid exploration](01_mine_explore.ipynb).

## Open the table

```python
import pandas as pd
from rdfsolve.client_api import Client

data = Client.open("../data/aopwikirdf.schema.json")
table = pd.read_csv("thyroid-chemicals.csv")
table
```

## Say which columns to use

“Identifier” identifies each chemical. “Name” supplies its title.

```python
chemicals = data.from_table(
    "Chemical entity", table, id_column="Identifier", title="Name"
)
chemicals
```

## Save the records

```python
data.save("chemicals-from-table.ttl", chemicals)
data.close()
```

## Look up a chemical in the new file

```python
local = Client.from_session(
    "thyroid-session.json", data_file="chemicals-from-table.ttl"
)
local.find("Phenobarbital")
```
